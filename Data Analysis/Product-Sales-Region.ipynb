```python
# Total Sales Analysis
df[['TotalPrice', 'Quantity']].agg(['sum', 'mean', 'min', 'max']).round(2)
```

```python
# Product Analysis
df.groupby('Product')[['TotalPrice', 'Quantity']].agg(['sum', 'mean', 'count']).round(2)
```

```python
# Region Analysis
df.groupby('Region')[['TotalPrice', 'Quantity']].agg(['sum', 'mean', 'count']).round(2)
```

```python
# Bar Chart - Total Sales by Product
fig, ax = plt.subplots(figsize=(4,2))
product_sales = df.groupby('Product')['TotalPrice'].sum()
ax.bar(product_sales.index, product_sales.values, color='skyblue', edgecolor='navy')
ax.set_title('Total Sales by Product', fontsize=14, fontweight='bold')
ax.set_xlabel('Product', fontsize=12)
ax.set_ylabel('Sales', fontsize=12)
ax.tick_params(axis='x', rotation=45)


plt.show()
```

```python
# Pie Chart - Sales Distribution by Region
plt.subplots(figsize=(6, 4))
df.groupby('Region')['TotalPrice'].sum().plot(kind='pie', ax=ax, autopct='%1.1f%%', startangle=90, )
ax.set_title('Sales Distribution by Region', fontsize=14, fontweight='bold')
ax.set_ylabel('')
plt.tight_layout()
plt.show()
```

```python
# Line Plot - Cumulative Sales by Region
fig, ax = plt.subplots(figsize=(6, 4))
for region in df['Region'].unique():
    region_data = df[df['Region'] == region].sort_values(by='TotalPrice')
    cumulative_sales = region_data['TotalPrice'].cumsum().values
    ax.plot(range(len(cumulative_sales)), cumulative_sales, marker='o', label=region, linewidth=2)
ax.set_title(' Sales Trend by Region', fontsize=14, fontweight='bold')
ax.set_xlabel('Order Index', fontsize=12)
ax.set_ylabel('Cumulative Sales ($)', fontsize=12)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot - Total Price vs Quantity (colored by Product)
fig, ax = plt.subplots(figsize=(6, 2))
colors = pd.factorize(df['Product'])[0]
scatter = ax.scatter(df['Quantity'], df['TotalPrice'], alpha=0.6, s=100, c=colors, cmap='viridis', edgecolors='black')
ax.set_title('Quantity vs Total Price by Product', fontsize=14, fontweight='bold')
ax.set_xlabel('Quantity', fontsize=12)
ax.set_ylabel('Total Price ($)', fontsize=12)
ax.grid(True, alpha=0.3)
plt.colorbar(scatter, ax=ax, label='Product')
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot - Total Price vs Quantity (colored by Product)
fig, ax = plt.subplots(figsize=(6, 4))
colors = pd.factorize(df['Product'])[0]
scatter = ax.scatter(df['Quantity'], df['TotalPrice'], alpha=0.6, s=100, c=colors, cmap='viridis', edgecolors='black')
ax.set_title('Quantity vs Total Price by Product', fontsize=14, fontweight='bold')
ax.set_xlabel('Quantity', fontsize=12)
ax.set_ylabel('Total Price ($)', fontsize=12)
ax.grid(True, alpha=0.3)
plt.colorbar(scatter, ax=ax, label='Product')
plt.tight_layout()
plt.show()
```

```python
# Business Insights
insights = {
    'Top Product (Sales)': df.groupby('Product')['TotalPrice'].sum().idxmax(),
    'Highest Sales Amount': f"${df.groupby('Product')['TotalPrice'].sum().max():,.2f}",
    'Top Region (Sales)': df.groupby('Region')['TotalPrice'].sum().idxmax(),
    'Total Revenue': f"${df['TotalPrice'].sum():,.2f}",
    'Average Order Value': f"${df['TotalPrice'].mean():,.2f}",
    'Total Quantity Sold': int(df['Quantity'].sum()),
    'Number of Regions': df['Region'].nunique(),
    'Number of Products': df['Product'].nunique()
}

pd.DataFrame(list(insights.items()), columns=['Metric', 'Value'])
```
